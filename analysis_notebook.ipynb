# Plotting Distribution of round winners by map

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

sns.countplot(data=df[["map", "round_winner"]], x='map', hue='round_winner')
plt.title('Distribution of Round Winners by Map')
plt.xlabel('Map')
plt.ylabel('Round winner')
plt.legend(title='Round Winner') 
plt.xticks(rotation=45)
plt.show()
```

```python
# Create the contingency table
contingency_table = pd.crosstab(df['map'], df['round_winner'])

contingency_table
```

## Hypothesis Test: Association Between map and round_winner

##### Null Hypothesis (H0): There is no association between map and round winner—the distribution of winners is the same for every map.

##### Alternative Hypothesis (H1): There is an association—the distribution of winners depends on the map

```python
# we will use chi square test the null vs alternate hypothesis
from scipy.stats import chi2_contingency

chi2, p, dof, expected = chi2_contingency(contingency_table)
print(f"Chi-square statistic: {chi2}, p-value: {p}, dof: {dof}")
```

### Inference
Since the p-value ≪ 0.05, we reject the null hypothesis (H₀) 

There is very strong statistical evidence of an association between the map and round winner — meaning the likelihood of winning depends on the map being played

```python
# since expected values are in the form of an array, lets convert it to a pandas df
expected_df = pd.DataFrame(expected, index=contingency_table.index, columns=contingency_table.columns)
print(expected_df)
```

```python
# lets get the residuals so that we can analyze which maps deviate the most 
std_residuals = (contingency_table - expected_df) / np.sqrt(expected_df)
print(std_residuals)
```

```python
# lets plot the residuals using a heatmap

sns.heatmap(std_residuals, annot=True, center=0, cmap="RdBu")
plt.title("Standardized Residuals: Map vs Side")
plt.xlabel("Side")
plt.ylabel("Map")
plt.show()
```

##### Maps with the largest (absolute) standardized residuals deviate most: In our data, maps like de_nuke for CT, and de_inferno/de_dust2 for T, stand out.

```python
# lets plot the observed vs expected round wins by map 
```

```python
df_plot = contingency_table.copy()
df_plot["CT_expected"] = expected[:, 0]
df_plot["T_expected"] = expected[:, 1]
df_plot[["CT", "CT_expected", "T", "T_expected"]].plot(kind="bar")
plt.title("Observed vs Expected Round Wins by Map")
plt.ylabel("Number of Rounds")
plt.show()
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import roc_curve
import numpy as np

def plot_ks_curve(y_true, y_proba):
    # Calculate False Positive Rate (FPR), True Positive Rate (TPR), and thresholds
    fpr, tpr, thresholds = roc_curve(y_true, y_proba)
    
    # Calculate KS statistic and find the threshold where KS is maximum
    ks_statistic = np.max(np.abs(tpr - fpr))
    ks_idx = np.argmax(np.abs(tpr - fpr))
    ks_threshold = thresholds[ks_idx]

    # Plot KS curve: TPR and FPR as cumulative distribution functions
    plt.figure(figsize=(8, 6))
    plt.plot(thresholds, tpr, label='TPR (Cumulative Distribution - Events)')
    plt.plot(thresholds, fpr, label='FPR (Cumulative Distribution - Non-Events)')
    plt.axvline(x=ks_threshold, color='red', linestyle='--', label=f'KS Threshold = {ks_threshold:.3f}')
    plt.gca().invert_xaxis()  # Show threshold descending (typical in scorecards)
    plt.xlabel('Threshold')
    plt.ylabel('Cumulative Distribution')
    plt.title(f'KS Curve (KS Statistic = {ks_statistic:.3f})')
    plt.legend()
    plt.grid(True)
    plt.show()

# Usage example:
# y_true = true binary labels (0/1)
# y_proba = predicted probabilities for positive class from your logistic regression

plot_ks_curve(y_test, y_proba)
```
